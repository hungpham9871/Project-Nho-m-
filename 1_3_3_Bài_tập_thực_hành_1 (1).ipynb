```python
# Tế bào 1: Cài đặt thư viện (Chạy riêng một lần)
# !pip install autoviz --upgrade --force-reinstall

# Tế bào 2: Mã nguồn AutoViz
import os, warnings
import pandas as pd
import matplotlib.pyplot as plt

# Đảm bảo biểu đồ hiển thị inline trên Colab
%matplotlib inline

warnings.filterwarnings("ignore")

# ── VÁ LỖI PANDAS 2.1+ CHO AUTOVIZ ────────────────────────
if not hasattr(pd.DataFrame, 'applymap'):
    pd.DataFrame.applymap = pd.DataFrame.map
# ──────────────────────────────────────────────────────────

from autoviz.AutoViz_Class import AutoViz_Class

# ── 1. Chuẩn bị dữ liệu ─────────────────────────────────────
# Đảm bảo bạn đã upload file 'marketing_campaign.csv' lên Colab
df_raw = pd.read_csv("marketing_campaign.csv", sep=';')
df = df_raw.copy()

df["Age"]          = 2024 - df["Year_Birth"]
s_cols         = ["MntWines","MntFruits","MntMeatProducts",
                       "MntFishProducts","MntSweetProducts","MntGoldProds"]
df["TotalSpend"]   = df[s_cols].sum(axis=1)
df["TotalAccepted"]= df[["AcceptedCmp1","AcceptedCmp2","AcceptedCmp3",
                           "AcceptedCmp4","AcceptedCmp5"]].sum(axis=1)
df["HasChildren"]  = ((df["Kidhome"] + df["Teenhome"]) > 0).astype(int)

drop_cols = ["ID","Dt_Customer","Year_Birth",
             "AcceptedCmp1","AcceptedCmp2","AcceptedCmp3","AcceptedCmp4","AcceptedCmp5"]
df_clean  = df.drop(columns=drop_cols)

# Chuyển đổi các cột object sang category
for col in df_clean.select_dtypes(include=['object']).columns:
    df_clean[col] = df_clean[col].astype('category')

print("="*60)
print("  MARKETING CAMPAIGN – AUTOVIZ EDA (COLAB DISPLAY)")
print("="*60)
print(f"  Dữ liệu: {df_clean.shape[0]:,} hàng × {df_clean.shape[1]} cột")

# Khởi tạo AutoViz
AV = AutoViz_Class()

# ══════════════════════════════════════════════════════════
# TIẾN HÀNH AUTOVIZ (Hiển thị trực tiếp)
# Thay verbose=2 thành verbose=1 và bỏ save_plot_dir
# ══════════════════════════════════════════════════════════

print("\n[1/5] AutoViz – EDA không depVar")
# Ghi chú: Có thể truyền thẳng DataFrame qua tham số dfte thay vì xuất ra CSV
dft1 = AV.AutoViz(filename="", sep=",", depVar="", dfte=df_clean, header=0,
                  verbose=1, lowess=False, chart_format="png",
                  max_rows_analyzed=2240, max_cols_analyzed=30)

print("\n[2/5] AutoViz – depVar='Response' (Binary Classification)")
dft2 = AV.AutoViz(filename="", sep=",", depVar="Response", dfte=df_clean, header=0,
                  verbose=1, lowess=False, chart_format="png",
                  max_rows_analyzed=2240, max_cols_analyzed=30)

print("\n[3/5] AutoViz – depVar='TotalSpend' (Regression) + lowess=True")
dft3 = AV.AutoViz(filename="", sep=",", depVar="TotalSpend", dfte=df_clean, header=0,
                  verbose=1, lowess=True, chart_format="png",
                  max_rows_analyzed=2240, max_cols_analyzed=30)

print("\n[4/5] AutoViz – Tập trung vào một số cột (SVG Format)")
focus_cols = ["Age","Income","TotalSpend","Recency","NumWebPurchases",
              "NumStorePurchases","Education","Marital_Status","HasChildren","Response"]
df_focus = df_clean[focus_cols].copy()

dft4 = AV.AutoViz(filename="", sep=",", depVar="Response", dfte=df_focus, header=0,
                  verbose=1, lowess=False, chart_format="svg",
                  max_rows_analyzed=2240, max_cols_analyzed=15)

print("\n[5/5] AutoViz – Tốc độ nhanh (Giới hạn max_rows=600, max_cols=8)")
dft5 = AV.AutoViz(filename="", sep=",", depVar="TotalSpend", dfte=df_clean, header=0,
                  verbose=1, lowess=False, chart_format="png",
                  max_rows_analyzed=600, max_cols_analyzed=8)

print("\n" + "="*60)
print("  HOÀN THÀNH AUTOVIZ EDA")
print("="*60)
```
